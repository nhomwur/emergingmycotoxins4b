```python
"""
ml_comparison_by_crop.py
=========================
Binary detection model for detectedC (median LOQ threshold per contaminant) at sample level.
Contaminant detection flags (was contaminant X detected above median LOQ in this sample?)
are added as features alongside weather columns.

Separate models are run for wheat, barley, and maize.
Validation: harvest_year == 2023 (held out entirely)
Train/test:  2019-2022, stratified 80/20
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.metrics import roc_auc_score, classification_report, RocCurveDisplay
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

MERGED_PATH  = "sample_level_dataset.csv"
RANDOM_STATE = 42
CROPS        = ["wheat", "barley", "maize"]

# -- 1. LOAD MERGED (contaminant-level) ------------------------------------
merged = pd.read_csv(MERGED_PATH, low_memory=False)
print(f"Loaded merged: {merged.shape}")

weather_cols = [c for c in merged.columns
                if c.startswith("month_") or c.startswith("week_")]

def fix_dutch_numeric_col(series):
    if series.dtype not in (object, "string"):
        return series
    def clean(v):
        if pd.isna(v): return v
        parts = str(v).strip().split(".")
        return parts[0] + "." + parts[1] if len(parts) > 2 else str(v)
    return pd.to_numeric(series.apply(clean), errors="coerce")

for col in weather_cols:
    merged[col] = fix_dutch_numeric_col(merged[col])

# -- 2. BUILD SAMPLE-LEVEL DATASET WITH CONTAMINANT FLAGS -----------------
def build_crop_dataset(df, crop):
    crop_df = df[df["product_harmonised"] == crop].copy()

    sample_cols = ["sample_id", "samplecountry", "harvest_year",
                   "product_harmonised", "detectedC"] + weather_cols
    sample_df = (
        crop_df[sample_cols]
        .groupby("sample_id")
        .agg({**{c: "first" for c in sample_cols if c not in ("sample_id", "detectedC")},
              "detectedC": "max"})
        .reset_index()
    )
    flag_cols = []  # no contaminant flags
    return sample_df, flag_cols


# -- 3. MODEL DEFINITIONS --------------------------------------------------
def make_pipeline(estimator, scale=True):
    steps = [("imputer", SimpleImputer(strategy="median"))]
    if scale:
        steps.append(("scaler", StandardScaler()))
    steps.append(("model", estimator))
    return Pipeline(steps)


# -- 4. RUN PER CROP -------------------------------------------------------
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
all_results = []

# For ROC plots: one figure per crop
fig_roc, axes_roc = plt.subplots(len(CROPS), 2,
                                  figsize=(14, 5 * len(CROPS)))

for crop_idx, crop in enumerate(CROPS):
    print(f"\n{'='*60}")
    print(f"CROP: {crop.upper()}")
    print(f"{'='*60}")

    sample_df, flag_cols = build_crop_dataset(merged, crop)
    FEATURES = weather_cols
    TARGET   = "detectedC"

    print(f"Sample-level shape: {sample_df.shape}")
    print(f"Contaminant flag features: {len(flag_cols)}")
    print(f"Flags: {flag_cols}")

    # Split
    val_df       = sample_df[sample_df["harvest_year"] == 2023].copy()
    traintest_df = sample_df[sample_df["harvest_year"] != 2023].copy()

    print(f"\nTrain+test: {len(traintest_df)}  Val (2023): {len(val_df)}")
    print(f"detectedC rate — train: {traintest_df[TARGET].mean():.2f}  "
          f"val: {val_df[TARGET].mean():.2f}")

    if val_df[TARGET].sum() < 10:
        print(f"WARNING: only {val_df[TARGET].sum()} positives in validation — "
              f"results unreliable for {crop}")

    X_traintest = traintest_df[FEATURES]
    y_traintest = traintest_df[TARGET]
    X_val       = val_df[FEATURES]
    y_val       = val_df[TARGET]

    X_train, X_test, y_train, y_test = train_test_split(
        X_traintest, y_traintest,
        test_size=0.2, stratify=y_traintest, random_state=RANDOM_STATE,
    )

    pos_weight = (y_train == 0).sum() / max((y_train == 1).sum(), 1)

    MODELS = {
        "Logistic Regression": make_pipeline(
            LogisticRegression(class_weight="balanced", max_iter=1000,
                               random_state=RANDOM_STATE)
        ),
        "Naive Bayes": make_pipeline(GaussianNB()),
        "KNN": make_pipeline(KNeighborsClassifier(n_neighbors=15)),
        "Random Forest": make_pipeline(
            RandomForestClassifier(n_estimators=500, max_depth=10,
                                   min_samples_leaf=5, class_weight="balanced",
                                   random_state=RANDOM_STATE, n_jobs=-1),
            scale=False,
        ),
        "XGBoost": make_pipeline(
            XGBClassifier(n_estimators=500, max_depth=5, learning_rate=0.05,
                          scale_pos_weight=pos_weight, random_state=RANDOM_STATE,
                          eval_metric="logloss", verbosity=0),
            scale=False,
        ),
        "LightGBM": make_pipeline(
            LGBMClassifier(n_estimators=500, max_depth=5, learning_rate=0.05,
                           class_weight="balanced", random_state=RANDOM_STATE,
                           verbosity=-1),
            scale=False,
        ),
        "SVM": make_pipeline(
            SVC(kernel="rbf", class_weight="balanced", probability=True,
                random_state=RANDOM_STATE)
        ),
    }

    crop_results = []
    for name, pipeline in MODELS.items():
        print(f"\n  Fitting {name}...")
        cv_auc = cross_val_score(pipeline, X_train, y_train,
                                 cv=cv, scoring="roc_auc").mean()
        pipeline.fit(X_train, y_train)

        test_proba = pipeline.predict_proba(X_test)[:, 1]
        val_proba  = pipeline.predict_proba(X_val)[:, 1]
        test_auc   = roc_auc_score(y_test, test_proba)
        val_auc    = roc_auc_score(y_val,  val_proba)

        print(f"    CV AUC: {cv_auc:.3f}  Test AUC: {test_auc:.3f}  "
              f"Val AUC: {val_auc:.3f}")

        crop_results.append({
            "Crop":     crop,
            "Model":    name,
            "CV AUC":   round(cv_auc,  3),
            "Test AUC": round(test_auc, 3),
            "Val AUC":  round(val_auc,  3),
            "AUC gap":  round(test_auc - val_auc, 3),
        })
        all_results.append(crop_results[-1])

        # ROC curves
        RocCurveDisplay.from_predictions(
            y_test, test_proba, ax=axes_roc[crop_idx, 0], name=name)
        RocCurveDisplay.from_predictions(
            y_val, val_proba,   ax=axes_roc[crop_idx, 1], name=name)

    axes_roc[crop_idx, 0].set_title(f"{crop.capitalize()} — ROC Test set")
    axes_roc[crop_idx, 1].set_title(f"{crop.capitalize()} — ROC Validation 2023")
    for ax in axes_roc[crop_idx]:
        ax.legend(fontsize=6)

    # Per-crop summary
    crop_df_results = pd.DataFrame(crop_results).sort_values("Val AUC", ascending=False)
    print(f"\n  {crop.upper()} SUMMARY:")
    print(crop_df_results.to_string(index=False))

    # Classification report for best model
    best_name = crop_df_results.iloc[0]["Model"]
    best_pipe = MODELS[best_name]
    print(f"\n  Best model: {best_name}")
    print(f"  --- TEST SET ---")
    print(classification_report(y_test, best_pipe.predict(X_test),
                                target_names=["not detected", "detected"]))
    print(f"  --- VALIDATION 2023 ---")
    print(classification_report(y_val, best_pipe.predict(X_val),
                                target_names=["not detected", "detected"]))

# -- 5. SAVE ROC PLOT ------------------------------------------------------
plt.tight_layout()
plt.savefig("model_comparison_by_crop_roc_detectedC.png", dpi=150)
plt.show()
print("Plot saved -> model_comparison_by_crop_roc_detectedC.png")

# -- 6. OVERALL SUMMARY TABLE ----------------------------------------------
print("\n" + "="*60)
print("OVERALL SUMMARY — all crops, sorted by Crop + Val AUC")
print("="*60)
all_df = pd.DataFrame(all_results).sort_values(
    ["Crop", "Val AUC"], ascending=[True, False])
print(all_df.to_string(index=False))

# -- 7. BAR CHART ----------------------------------------------------------
fig, axes_bar = plt.subplots(1, len(CROPS), figsize=(6 * len(CROPS), 5))
for ax, crop in zip(axes_bar, CROPS):
    crop_data = all_df[all_df["Crop"] == crop].sort_values("Val AUC", ascending=False)
    x = np.arange(len(crop_data))
    width = 0.25
    ax.bar(x - width, crop_data["CV AUC"],  width, label="CV AUC")
    ax.bar(x,         crop_data["Test AUC"], width, label="Test AUC")
    ax.bar(x + width, crop_data["Val AUC"],  width, label="Val AUC (2023)")
    ax.set_xticks(x)
    ax.set_xticklabels(crop_data["Model"], rotation=25, ha="right", fontsize=8)
    ax.set_ylim(0.4, 1.0)
    ax.set_ylabel("ROC-AUC")
    ax.set_title(f"{crop.capitalize()} — model comparison")
    ax.legend(fontsize=7)
    ax.axhline(0.5, color="grey", linestyle="--", linewidth=0.8)

plt.tight_layout()
plt.savefig("model_comparison_by_crop_bar_detectedC.png", dpi=150)
plt.show()
print("Plot saved -> model_comparison_by_crop_bar_detectedC.png")
```
